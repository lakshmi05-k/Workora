```python

import pandas as pd

df = pd.read_csv('/content/customer_churn_raw_v2.csv')

print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python

clean_df = df.copy()

clean_df = clean_df.drop_duplicates()

clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.replace(" ", "_")
)

text_columns = clean_df.select_dtypes(
    include="object"
).columns

for col in text_columns:
    clean_df[col] = clean_df[col].str.strip()

clean_df["TotalChargesINR"] = pd.to_numeric(
    clean_df["TotalChargesINR"],
    errors="coerce"
)

print("Original dataset shape:", df.shape)
print("After removing duplicates:", clean_df.shape)

print("\nMissing values after initial cleaning:")
print(clean_df.isnull().sum())

print("\nUpdated data types:")
print(clean_df.dtypes)
```

```python

clean_df["Region"] = clean_df["Region"].fillna("Unknown")

mask = clean_df["TotalChargesINR"].isna()

clean_df.loc[mask, "TotalChargesINR"] = (
    clean_df.loc[mask, "MonthlyChargesINR"]
    * clean_df.loc[mask, "TenureMonths"]
)

print("Missing values after handling:")
print(clean_df.isnull().sum())

print("\nFinal dataset shape:", clean_df.shape)
```

```python

clean_df.to_csv(
    "customer_churn_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
print("Rows:", clean_df.shape[0])
print("Columns:", clean_df.shape[1])
print("Missing values:", clean_df.isnull().sum().sum())
print("Duplicate rows:", clean_df.duplicated().sum())

from google.colab import files
files.download("customer_churn_cleaned.csv")
```
